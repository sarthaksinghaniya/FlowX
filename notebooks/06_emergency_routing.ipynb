## Step 2. Routing Logic

This notebook uses a simple breadth-first search over a small road graph. Replace it with a richer graph or live road-state data when available.

```python
from collections import deque

ROAD_GRAPH = {
    'A': ['B'],
    'B': ['C'],
    'C': ['D'],
    'D': [],
}


def compute_route(start: str, end: str) -> list[str]:
    queue = deque([(start, [start])])
    visited = {start}
    while queue:
        node, path = queue.popleft()
        if node == end:
            return path
        for neighbor in ROAD_GRAPH.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))
    return []

route = compute_route('A', 'D')
route_df = pd.DataFrame({'sequence': range(1, len(route) + 1), 'node': route})
route_df.to_csv(ROUTE_CSV, index=False)
print(f'Saved route to {ROUTE_CSV}')
route_df
```
