## Step 1. Setup

```python
import os
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
OUTPUT_ROOT = PROJECT_ROOT / 'outputs'
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

print(f'Project root: {PROJECT_ROOT}')
print(f'Outputs root: {OUTPUT_ROOT}')

OUTPUT_DIR = OUTPUT_ROOT / 'predictive_congestion'
os.makedirs(OUTPUT_DIR, exist_ok=True)
PREDICTION_CSV = OUTPUT_DIR / 'predicted_congestion.csv'
PREDICTION_PLOT = OUTPUT_DIR / 'predicted_congestion.png'
```

## Step 2. Placeholder Forecasting Logic

This notebook intentionally uses a lightweight synthetic sequence. Replace it with a real LSTM, temporal GNN, or RL environment when your training assets are ready.

```python
time_steps = np.arange(0, 12)
historical_density = np.array([0.22, 0.25, 0.31, 0.40, 0.46, 0.50, 0.48, 0.44, 0.38, 0.35, 0.30, 0.28])


def predict_congestion(sequence: np.ndarray, horizon: int = 6) -> pd.DataFrame:
    baseline = sequence[-1]
    trend = np.linspace(0.02, 0.10, horizon)
    predicted = np.clip(baseline + trend, 0.0, 1.0)
    return pd.DataFrame({
        'future_step': np.arange(1, horizon + 1),
        'predicted_density': predicted,
    })

prediction_df = predict_congestion(historical_density)
prediction_df.to_csv(PREDICTION_CSV, index=False)
print(f'Saved predictions to {PREDICTION_CSV}')
prediction_df
```

## Step 3. Save Forecast Plot

```python
plt.figure(figsize=(8, 4))
plt.plot(time_steps, historical_density, marker='o', label='Historical')
plt.plot(
    np.arange(time_steps[-1] + 1, time_steps[-1] + 1 + len(prediction_df)),
    prediction_df['predicted_density'],
    marker='o',
    linestyle='--',
    label='Predicted',
)
plt.title('Predictive Congestion Forecast')
plt.xlabel('Time Step')
plt.ylabel('Density')
plt.legend()
plt.tight_layout()
plt.savefig(PREDICTION_PLOT, dpi=150)
plt.show()
print(f'Saved plot to {PREDICTION_PLOT}')
```
