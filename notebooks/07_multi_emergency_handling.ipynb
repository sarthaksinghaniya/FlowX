# 07 Multi Emergency Handling

**Description:** Explore logic for multiple simultaneous emergency requests competing for signal priority.

**Objective:** Save prioritization decisions and a simple comparison plot into `outputs/multi_emergency_handling/`.

## Step 1. Setup

```python
import os
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
OUTPUT_ROOT = PROJECT_ROOT / 'outputs'
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

print(f'Project root: {PROJECT_ROOT}')
print(f'Outputs root: {OUTPUT_ROOT}')

OUTPUT_DIR = OUTPUT_ROOT / 'multi_emergency_handling'
os.makedirs(OUTPUT_DIR, exist_ok=True)
DECISION_CSV = OUTPUT_DIR / 'multi_emergency_decisions.csv'
DECISION_PLOT = OUTPUT_DIR / 'multi_emergency_priority.png'
```

## Step 2. Priority Resolution

Replace the scoring logic with your final policy engine, RL agent, or dispatch constraints.

```python
requests_df = pd.DataFrame([
    {'vehicle_id': 'EV-1', 'eta_sec': 18, 'severity': 5},
    {'vehicle_id': 'EV-2', 'eta_sec': 35, 'severity': 4},
    {'vehicle_id': 'EV-3', 'eta_sec': 22, 'severity': 3},
])


def prioritize_emergencies(df: pd.DataFrame) -> pd.DataFrame:
    ranked = df.copy()
    ranked['priority_score'] = (ranked['severity'] * 10) - ranked['eta_sec']
    ranked = ranked.sort_values('priority_score', ascending=False).reset_index(drop=True)
    ranked['decision_rank'] = ranked.index + 1
    return ranked

priority_df = prioritize_emergencies(requests_df)
priority_df.to_csv(DECISION_CSV, index=False)
print(f'Saved decision log to {DECISION_CSV}')
priority_df
```

## Step 3. Save Plot

```python
plt.figure(figsize=(8, 4))
plt.bar(priority_df['vehicle_id'], priority_df['priority_score'], color='crimson')
plt.title('Emergency Priority Scores')
plt.xlabel('Emergency Vehicle')
plt.ylabel('Priority Score')
plt.tight_layout()
plt.savefig(DECISION_PLOT, dpi=150)
plt.show()
print(f'Saved plot to {DECISION_PLOT}')
```
