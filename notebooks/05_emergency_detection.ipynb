# 05 Emergency Detection

**Description:** Build a placeholder emergency detection workflow using audio/video cues such as sirens or marked emergency vehicles.

**Objective:** Save emergency event logs and annotated frames into `outputs/emergency_detection/`.

## Step 1. Setup

```python
import os
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
OUTPUT_ROOT = PROJECT_ROOT / 'outputs'
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

print(f'Project root: {PROJECT_ROOT}')
print(f'Outputs root: {OUTPUT_ROOT}')

OUTPUT_DIR = OUTPUT_ROOT / 'emergency_detection'
os.makedirs(OUTPUT_DIR, exist_ok=True)
LOG_CSV = OUTPUT_DIR / 'emergency_detection_log.csv'
FRAME_PATH = OUTPUT_DIR / 'emergency_frame_annotated.jpg'
```

## Step 2. Placeholder Signal Detection

Replace the heuristic below with actual siren classification, audio embeddings, or visual emergency vehicle detection logic.

```python
def detect_emergency_signals(audio_level: float, visual_score: float, threshold: float = 0.7) -> dict:
    combined_score = (audio_level * 0.5) + (visual_score * 0.5)
    return {
        'audio_level': audio_level,
        'visual_score': visual_score,
        'combined_score': combined_score,
        'emergency_detected': combined_score >= threshold,
    }

result = detect_emergency_signals(audio_level=0.85, visual_score=0.75)
log_df = pd.DataFrame([result])
log_df.to_csv(LOG_CSV, index=False)
print(f'Saved emergency log to {LOG_CSV}')
log_df
```

## Step 3. Save Annotated Frame

```python
frame = np.full((480, 640, 3), 30, dtype=np.uint8)
label = 'Emergency Detected' if result['emergency_detected'] else 'No Emergency'
color = (0, 0, 255) if result['emergency_detected'] else (0, 180, 0)
cv2.putText(frame, label, (140, 240), cv2.FONT_HERSHEY_SIMPLEX, 1.0, color, 3)
cv2.imwrite(str(FRAME_PATH), frame)

frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 5))
plt.imshow(frame_rgb)
plt.title('Emergency Detection Output')
plt.axis('off')
plt.show()
print(f'Saved annotated frame to {FRAME_PATH}')
```
